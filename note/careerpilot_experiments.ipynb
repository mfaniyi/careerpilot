```python
from pydantic import BaseModel, Field


# Represents the user's career background and current situation
class CareerProfile(BaseModel):
    name: str
    current_role: str
    years_of_experience: int = Field(ge=0)
    education: list[str]
    skills: list[str]
    certifications: list[str]
    industry: str
    career_goal: str


# Represents a technology career that CareerPilot can recommend
class CareerPath(BaseModel):
    role: str
    description: str
    required_skills: list[str]
    preferred_skills: list[str]
    entry_level: str
    related_roles: list[str]
```

```python
# Create a validated career profile
candidate = CareerProfile(
    name="Test User",
    current_role="Accountant",
    years_of_experience=6,
    education=["BSc Accounting"],
    skills=[
        "Excel",
        "Financial Reporting",
        "Communication"
    ],
    certifications=[],
    industry="Finance",
    career_goal="Business Analyst",
)


# Create a validated career path
business_analyst = CareerPath(
    role="Business Analyst",
    description="Analyzes business processes and requirements to support better decisions.",
    required_skills=[
        "Requirements Analysis",
        "Communication",
        "Process Modelling",
        "SQL",
    ],
    preferred_skills=[
        "Power BI",
        "Excel",
    ],
    entry_level="Beginner",
    related_roles=[
        "Data Analyst",
        "Product Analyst",
    ],
)

print(candidate)
print()
print(business_analyst)
```
